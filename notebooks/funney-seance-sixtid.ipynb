```python
from pathlib import Path

candidate_paths = [
    Path('funney-seance-sixtid.csv.csv'),
    Path(r'c:\Users\laura\AppData\Local\Temp\funney-seance-sixtid.csv.csv'),
]

CSV_PATH = next((path for path in candidate_paths if path.exists()), None)
if CSV_PATH is None:
    raise FileNotFoundError(
        'CSV not found. Checked the notebook folder and the temp file path; edit candidate_paths if needed.'
    )

CSV_PATH
```

```python
import pandas as pd

df = pd.read_csv(
    CSV_PATH,
    header=None,
    names=['seed', 'total', 'first', 'second', 'third'],
)

df['imbalance'] = (df[['first', 'second', 'third']].max(axis=1) - df[['first', 'second', 'third']].min(axis=1))
df['all_three_nonzero'] = (df[['first', 'second', 'third']] > 0).all(axis=1)
df['selected'] = df['seed'].eq('GWHQAQ1L')

df.head(12)
```

```python
summary = {
    'row_count': int(len(df)),
    'selected_seed_rows': df.loc[df['selected']].to_dict(orient='records'),
    'top_12_rows': df.head(12).to_dict(orient='records'),
    'total_distribution': df['total'].value_counts().sort_index(ascending=False).to_dict(),
}

summary
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

df['total'].value_counts().sort_index().plot(kind='bar', ax=axes[0], color='#4C78A8')
axes[0].set_title('Total score distribution')
axes[0].set_xlabel('total')
axes[0].set_ylabel('count')

(
    df.sort_values(['total', 'third', 'first', 'second'], ascending=False)
      .head(15)
      .set_index('seed')[['first', 'second', 'third']]
      .plot(kind='bar', stacked=True, ax=axes[1], color=['#59A14F', '#F28E2B', '#E15759'])
)
axes[1].set_title('Top 15 component mix')
axes[1].set_xlabel('seed')
axes[1].set_ylabel('component total')
axes[1].legend(loc='upper right')

plt.tight_layout()
plt.show()
```
